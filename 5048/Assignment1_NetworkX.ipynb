```python
# calculate the centrality of degree and select top sorted 150 nodes
degree_centrality = nx.degree_centrality(G)
top_150 = sorted(degree_centrality, key=lambda k: degree_centrality[k], reverse=True)[:150]

print(top_150)
```

```python
# Extract subgraph
subgraph_150 = G.subgraph(top_150)
```

```python
nx.draw_spring(subgraph_150, with_labels=True)
plt.show()
```

```python
plt.figure(figsize=(30, 30))
pos = nx.spring_layout(subgraph_150, k=1, iterations=100, seed=42)
nx.draw_networkx_nodes(subgraph_150, pos)
nx.draw_networkx_edges(subgraph_150, pos, alpha=0.3)
nx.draw_networkx_labels(subgraph_150, pos, font_size=8)
plt.title("Top 150 Important Borad Games")
plt.savefig("Borad Game 1")
plt.show()
```

```python
# To check the attributes of node
for node, attrs in list(G.nodes(data=True))[:5]:
    print("node is " + node + " -> ", attrs)
```

```python
# Extract nodes
power_node = [n for n, attr in G.nodes(data=True) if 'Variable Player Powers' in attr.get('mechanics', [])]
# print(power_node)
# create a new subgraph
power_subgraph = G.subgraph(power_node)
print(power_subgraph)
```

```python
# Modularity-based
partition = community_louvain.best_partition(power_subgraph)

# The number of community
len(set(partition.values()))
for node in power_subgraph.nodes():
    print(node)
# Assign community number colors to each node
community_colors = [partition[node] for node in power_subgraph.nodes()]
print(power_subgraph.nodes())
```

```python
# Draw the pic
plt.figure(figsize=(20, 20))
pos = nx.spring_layout(power_subgraph, k=1.5, iterations=90, seed=42)
nx.draw_networkx_nodes(power_subgraph, pos, cmap=plt.cm.tab20, node_color=community_colors)
nx.draw_networkx_edges(power_subgraph, pos, alpha=0.2)
nx.draw_networkx_labels(power_subgraph, pos, font_size=8)
plt.savefig("Topic of interest")
plt.show()
```

```python
from pyvis.network import Network

net = Network(notebook=True, width='900px', height='900px')

# 从NetworkX加载图
net.from_nx(power_subgraph)

# 显示交互式网络
net.show("interactive_graph.html")
```

```python
from collections import defaultdict

# list of nodes for each community
community_groups = defaultdict(list)
for node, community_id in partition.items():
    community_groups[community_id].append(node)

# nodes for each community
for community_id, nodes in community_groups.items():
    print(f"Community {community_id}: {nodes}\n")
```
